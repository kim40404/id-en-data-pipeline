# Dataset Exploration and Cleaning

In this notebook, we will load the raw ID-EN bilingual dataset from Hugging Face and explore the dirty data.

```python
import polars as pl
from datasets import load_dataset

# Set display options for Polars
pl.Config.set_tbl_rows(10)

# Load raw dataset from Hugging Face
print("Mendownload dataset dari Hugging Face...")
hf_dataset = load_dataset("Kimsang766/agentic-ai-instructions-id-en", split="train")

# Convert to Polars DataFrame
df = pl.from_arrow(hf_dataset.data.table)

print(f"\nBerhasil memuat {df.shape[0]} baris data!")
display(df.head())
```

```python
# 1. Cek jumlah data kotor (Duplikat & Kosong)
print("--- SEBELUM DIBERSIHKAN ---")
print(f"Total baris awal: {len(df)}")
print(f"Jumlah duplikat persis: {df.is_duplicated().sum()}")
print("\nJumlah baris kosong (null) di tiap kolom:")
display(df.null_count())

# 2. Proses Pembersihan (Cleaning) menggunakan Polars
df_clean = (
    df
    .drop_nulls()  # Membuang semua baris yang kosong
    .unique()      # Membuang baris yang duplikat
)

# 3. Hasil setelah dibersihkan
print("\n--- SETELAH DIBERSIHKAN ---")
print(f"Total baris bersih: {len(df_clean)}")
print(f"Baris sampah yang dibuang: {len(df) - len(df_clean)}")
```
